# Singapore Salary Insights Dashboard
## Interactive Analysis of 1M+ Job Postings

**Business Objective:** Provide HR departments, consultancy firms, job seekers, and government agencies with actionable salary insights for benchmarking and planning.

**Dataset:** 1,044,583 cleaned job postings from 53,151 companies across multiple industries.


## Section 1: Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
import json
import warnings
warnings.filterwarnings('ignore')

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)

print("✓ All libraries imported successfully")
```

## Section 2: Load and Prepare Data

```python
# Load cleaned data
df = pd.read_csv('SGJobData_cleaned.csv')

# ── Salary Outlier Removal ─────────────────────────────────────────────────────────────────────────
# salary_minimum < 1200: placeholder/part-time/internship noise below Singapore LQS floor
# salary_maximum > 30000: annual salary entered in monthly field
# Decision: Drop (not impute) - no recoverable info. Impact: 8,781 rows (0.84%)
rows_before = len(df)
df = df[
    (df['salary_minimum'] >= 1200) &
    (df['salary_maximum'] <= 30_000)
].copy()
rows_dropped = rows_before - len(df)
print(f'Salary outliers dropped: {rows_dropped:,} rows ({rows_dropped/rows_before*100:.2f}%)')
print(f'Remaining records: {len(df):,}')

# Display dataset overview
print("=" * 80)
print("DATASET OVERVIEW")
print("=" * 80)
print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"Unique Companies: {df['postedCompany_name'].nunique():,}")
print(f"Unique Job Titles: {df['title'].nunique():,}")
print(f"\nSalary Statistics:")
print(f"  Min: SGD {df['salary_minimum'].min():,.0f}")
print(f"  Max: SGD {df['salary_maximum'].max():,.0f}")
print(f"  Mean: SGD {df['average_salary'].mean():,.2f}")
print(f"  Median: SGD {df['average_salary'].median():,.2f}")

# Prepare additional columns for analysis
df['salary_spread'] = df['salary_maximum'] - df['salary_minimum']
df['posting_date'] = pd.to_datetime(df['metadata_newPostingDate'], errors='coerce')
df['year_month'] = df['posting_date'].dt.to_period('M')

# Extract industries from JSON categories
# The 'categories' column contains JSON strings; we will parse them to extract the primary industry
# Example of a categories entry: '[{"category": "Information Technology"}, {"category": "Software Development"}]'
# We will take the first category as the primary industry
# Create a new column 'primary_industry'
# Parse categories and extract primary industry
# Initialize an empty list to hold industry data
industries_list = []
for idx, cat_str in enumerate(df['categories'].dropna()):
    try:
        categories = json.loads(cat_str)
        if isinstance(categories, list) and len(categories) > 0:
            if isinstance(categories[0], dict) and 'category' in categories[0]:
                industries_list.append({
                    'idx': idx,
                    'industry': categories[0]['category']
                })
    except:
        pass

if industries_list:
    ind_df = pd.DataFrame(industries_list).set_index('idx')
    # Map primary industry back to the main dataframe
    # Use index to align
    # Create a new column 'primary_industry' in df
    # Map industries
    df['primary_industry'] = df.index.map(lambda x: ind_df.loc[x, 'industry'] if x in ind_df.index else 'Other')
else:
    # If no industries were found, set all to 'Other'
    df['primary_industry'] = 'Other'

print(f"\n✓ Data loaded and prepared successfully")
```

## Section 3: Create Key Metrics & Visualizations

```python
# Chart 1: Salary Distribution (Histogram with KDE)
# KDE is computationally expensive; we will use only histogram here for performance
# KDE stands for Kernel Density Estimate, which is a way to estimate the probability density function of a continuous random variable.
# KDE functions used below, e.g. add_trace(go.Histogram(..., histnorm='density')) can be computationally expensive for large datasets.
try:
    # Use sampling for efficiency with large dataset
    sample_size = min(50000, len(df))
    df_sample = df.sample(n=sample_size, random_state=42)
    
    fig1 = go.Figure()
    fig1.add_trace(go.Histogram(
        x=df_sample['average_salary'],
        nbinsx=50,
        name='Salary Distribution',
        marker_color='rgba(0, 123, 255, 0.7)',
        hovertemplate='<b>Salary Range</b><br>SGD %{x:,.0f}<br><b>Count</b>: %{y}<extra></extra>'
    ))

    fig1.add_vline(x=df['average_salary'].median(), line_dash="dash", line_color="red", 
                  annotation_text=f"Median: SGD {df['average_salary'].median():,.0f}",
                  annotation_position="top right")
    fig1.add_vline(x=df['average_salary'].mean(), line_dash="dot", line_color="green",
                  annotation_text=f"Mean: SGD {df['average_salary'].mean():,.0f}",
                  annotation_position="top left")

    fig1.update_layout(
        title='<b>Salary Distribution across all jobs</b>',
        xaxis_title='Average Salary (SGD)',
        yaxis_title='Number of Job Postings',
        hovermode='x unified',
        template='plotly_white',
        height=500
    )
    fig1.show()
    print("✓ Chart 1 created successfully")
except Exception as e:
    print(f"⚠ Error creating Chart 1: {str(e)}")
```

```python
# Chart 2: Salary by Employment Type (Box Plot)
try:
    emp_type_stats = df.groupby('employmentTypes', observed=True)['average_salary'].agg(['mean', 'median', 'count']).sort_values('mean', ascending=False)

    fig2 = go.Figure()
    for emp_type in emp_type_stats.index:
        emp_data = df[df['employmentTypes'] == emp_type]['average_salary']
        fig2.add_trace(go.Box(
            y=emp_data,
            name=emp_type,
            boxmean=False,
            hovertemplate='<b>%{fullData.name}</b><br>Salary: SGD %{y:,.0f}<extra></extra>'
        ))

    fig2.update_layout(
        title='<b>Salary Distribution by Employment Type</b>',
        yaxis_title='Average Salary (SGD)',
        xaxis_title='Employment Type',
        hovermode='closest',
        template='plotly_white',
        height=500,
        showlegend=False
    )
    fig2.show()

    # Display employment type statistics
    print("\n✓ Chart 2 created successfully")
    print("\nEmployment Type Statistics:")
    print(emp_type_stats.round(2))
except Exception as e:
    print(f"⚠ Error creating Chart 2: {str(e)}")
```

```python
# Chart 3: Salary by Position Level
try:
    pos_level_stats = df.groupby('positionLevels', observed=True)['average_salary'].agg(['mean', 'median', 'count']).sort_values('mean', ascending=False)

    fig3 = go.Figure()
    fig3.add_trace(go.Bar(
        x=pos_level_stats.index,
        y=pos_level_stats['mean'],
        name='Mean Salary',
        marker_color='rgba(0, 123, 255, 0.7)',
        text=[f"SGD {val:,.0f}" for val in pos_level_stats['mean']],
        textposition='auto',
        hovertemplate='<b>%{x}</b><br>Mean: SGD %{y:,.0f}<br>Count: %{customdata:,}<extra></extra>',
        customdata=pos_level_stats['count'].values
    ))

    fig3.add_trace(go.Scatter(
        x=pos_level_stats.index,
        y=pos_level_stats['median'],
        name='Median Salary',
        mode='lines+markers',
        line_color='red',
        marker=dict(size=8),
        hovertemplate='<b>%{x}</b><br>Median: SGD %{y:,.0f}<extra></extra>'
    ))

    fig3.update_layout(
        title='<b>Salary by Position Level</b>',
        xaxis_title='Position Level',
        yaxis_title='Salary (SGD)',
        hovermode='x unified',
        template='plotly_white',
        height=500,
        xaxis_tickangle=-45
    )
    fig3.show()

    print("\n✓ Chart 3 created successfully")
    print("\nPosition Level Statistics:")
    print(pos_level_stats.round(2))
except Exception as e:
    print(f"⚠ Error creating Chart 3: {str(e)}")
```

```python
# Chart 4: Experience vs Salary - FIXED CODE
try:
    # Filter and aggregate data
    df_exp = df[df['minimumYearsExperience'] <= 12].copy()
    exp_stats = df_exp.groupby('minimumYearsExperience')['average_salary'].agg(['mean', 'count']).reset_index()
    
    # Rename columns for clarity
    exp_stats.columns = ['years_experience', 'avg_salary', 'job_count']
    
    # Filter to only include experience levels with sufficient data
    exp_stats = exp_stats[exp_stats['job_count'] > 100].reset_index(drop=True)

    fig4 = go.Figure()
    fig4.add_trace(go.Scatter(
        x=exp_stats['years_experience'],
        y=exp_stats['avg_salary'],
        mode='lines+markers',
        name='Average Salary',
        line=dict(color='green', width=3),
        marker=dict(size=8, color='darkgreen'),
        hovertemplate='<b>Experience:</b> %{x} years<br><b>Salary:</b> SGD %{y:,.0f}<br><b>Jobs:</b> %{customdata:,}<extra></extra>',
        customdata=exp_stats['job_count'].values
    ))

    fig4.update_layout(
        title='<b>Salary Growth by Years of Experience</b>',
        xaxis_title='Minimum Years of Experience',
        yaxis_title='Average Salary (SGD)',
        hovermode='x unified',
        template='plotly_white',
        height=500,
        xaxis=dict(dtick=1)
    )
    fig4.show()

    print("\n✓ Chart 4 created successfully")
    print("\nExperience vs Salary:")
    print(exp_stats.round(2))
except Exception as e:
    print(f"⚠ Error creating Chart 4: {str(e)}")
    import traceback
    traceback.print_exc()
```

```python
# Chart 5: Top 15 Hiring Companies
try:
    top_companies = df['postedCompany_name'].value_counts().head(15).reset_index()
    top_companies.columns = ['Company', 'Postings']
    company_salary = df.groupby('postedCompany_name')['average_salary'].mean()
    top_companies['Avg_Salary'] = top_companies['Company'].map(company_salary)

    fig5 = go.Figure()
    fig5.add_trace(go.Bar(
        x=top_companies['Postings'],
        y=top_companies['Company'],
        orientation='h',
        marker=dict(
            color=top_companies['Avg_Salary'],
            colorscale='Viridis',
            showscale=True,
            colorbar=dict(title='Avg Salary<br>(SGD)')
        ),
        text=[f"SGD {sal:,.0f}" for sal in top_companies['Avg_Salary']],
        textposition='auto',
        hovertemplate='<b>%{y}</b><br>Postings: %{x}<br>Avg Salary: SGD %{marker.color:,.0f}<extra></extra>'
    ))

    fig5.update_layout(
        title='<b>Top 15 Hiring Companies</b>',
        xaxis_title='Number of Job Postings',
        yaxis_title='Company Name',
        hovermode='closest',
        template='plotly_white',
        height=600,
        margin=dict(l=200)
    )
    fig5.show()
    print("✓ Chart 5 created successfully")
except Exception as e:
    print(f"⚠ Error creating Chart 5: {str(e)}")
```

```python
# Chart 6: Salary Percentiles
try:
    percentiles = [10, 25, 50, 75, 90]
    percentile_values = [np.percentile(df['average_salary'], p) for p in percentiles]

    fig6 = go.Figure()
    fig6.add_trace(go.Bar(
        x=[f"{p}th" for p in percentiles],
        y=percentile_values,
        marker_color=['#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b'],
        text=[f"SGD {val:,.0f}" for val in percentile_values],
        textposition='auto',
        hovertemplate='<b>%{x} Percentile</b><br>Salary: SGD %{y:,.0f}<extra></extra>'
    ))

    fig6.update_layout(
        title='<b>Salary Percentile Distribution</b>',
        xaxis_title='Percentile',
        yaxis_title='Salary (SGD)',
        hovermode='x unified',
        template='plotly_white',
        height=500
    )
    fig6.show()

    print("\n✓ Chart 6 created successfully")
    print("\nSalary Percentiles:")
    for p, val in zip(percentiles, percentile_values):
        print(f"  {p}th: SGD {val:,.2f}")
except Exception as e:
    print(f"⚠ Error creating Chart 6: {str(e)}")
```

```python
# Chart 7: Top Industries by Salary
try:
    # Get top industries safely
    if 'primary_industry' in df.columns and df['primary_industry'].notna().any():
        top_industries = df['primary_industry'].value_counts().head(10).index
        industry_stats = df[df['primary_industry'].isin(top_industries)].groupby('primary_industry', observed=True)['average_salary'].agg(['mean', 'count']).sort_values('mean', ascending=False)
    else:
        # Fallback: use employment types
        industry_stats = df.groupby('employmentTypes', observed=True)['average_salary'].agg(['mean', 'count']).sort_values('mean', ascending=False).head(10)
        industry_stats.index.name = 'Category'

    fig7 = go.Figure()
    fig7.add_trace(go.Bar(
        x=industry_stats.index,
        y=industry_stats['mean'],
        marker_color='rgba(0, 123, 255, 0.7)',
        text=[f"SGD {val:,.0f}<br>({int(cnt)} jobs)" for val, cnt in zip(industry_stats['mean'], industry_stats['count'])],
        textposition='auto',
        hovertemplate='<b>%{x}</b><br>Mean: SGD %{y:,.0f}<br>Count: %{customdata:,}<extra></extra>',
        customdata=industry_stats['count'].values
    ))

    fig7.update_layout(
        title='<b>Top Industries/Categories by Average Salary</b>',
        xaxis_title='Industry/Category',
        yaxis_title='Average Salary (SGD)',
        hovermode='x unified',
        template='plotly_white',
        height=500,
        xaxis_tickangle=-45
    )
    fig7.show()
    print("✓ Chart 7 created successfully")
except Exception as e:
    print(f"⚠ Error creating Chart 7: {str(e)}")
```

```python
# Chart 8: Employment Type Distribution (Pie Chart)
try:
    emp_counts = df['employmentTypes'].value_counts()

    fig8 = go.Figure(data=[go.Pie(
        labels=emp_counts.index,
        values=emp_counts.values,
        hovertemplate='<b>%{label}</b><br>Count: %{value:,}<br>Percentage: %{percent}<extra></extra>',
        textinfo='label+percent',
        textposition='inside'
    )])

    fig8.update_layout(
        title='<b>Market Distribution by Employment Type</b>',
        height=500,
        template='plotly_white'
    )
    fig8.show()

    print("\n✓ Chart 8 created successfully")
    print("\nEmployment Type Distribution:")
    print(emp_counts)
    print(f"\nPercentages:")
    print((emp_counts / emp_counts.sum() * 100).round(2))
except Exception as e:
    print(f"⚠ Error creating Chart 8: {str(e)}")
```
